# Training loss and hypothesis outcomes

Each bar's **total height is average training loss**, sorted from highest to lowest. Four equal-height sections encode hypothesis outcomes by color. From **bottom to top**: Q50 vs Q00 · H1, Q50 vs Q00 · H2, Q05 vs Q00 · H1, Q05 vs Q00 · H2.

Section heights are a visual subdivision of total loss, **not separate loss estimates for each hypothesis**. Green = success; red = failure; dark red = not estimable. White dividers and labels keep all four sections visible even when their outcomes match.

Data are transcribed from the supplied results. Edit the records below to update the figure.


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from pathlib import Path

# Outcomes follow the bottom-to-top section order below.
section_labels = [
    "Q50 vs Q00 · H1",
    "Q50 vs Q00 · H2",
    "Q05 vs Q00 · H1",
    "Q05 vs Q00 · H2",
]
records = [
    {"group": "3 mechanisms", "loss": 2.27,
     "outcomes": ["success", "success", "success", "success"]},
    {"group": "3 mechanisms\n(one altered)", "loss": 6.23,
     "outcomes": ["not estimable"] * 4},
    {"group": "2 mechanisms\n(12)", "loss": 6.20,
     "outcomes": ["success", "success", "failure", "success"]},
    {"group": "2 mechanisms\n(13)", "loss": 7.43,
     "outcomes": ["not estimable"] * 4},
    {"group": "2 mechanisms\n(23)", "loss": 2.90,
     "outcomes": ["success", "success", "success", "success"]},
]
colors = {"success": "#C2E5CF", "failure": "#F2B8AE", "not estimable": "#DD7389"}
records = sorted(records, key=lambda row: row["loss"], reverse=True)
assert all(len(row["outcomes"]) == 4 for row in records)
assert all(outcome in colors for row in records for outcome in row["outcomes"])
```

```python
with plt.rc_context({"font.family": "DejaVu Sans", "font.size": 11}):
    fig, ax = plt.subplots(figsize=(13, 7.5))
    fig.patch.set_facecolor("white")
    ax.set_facecolor("white")
    for x, row in enumerate(records):
        section_height = row["loss"] / 4
        for section, (label, outcome) in enumerate(zip(section_labels, row["outcomes"])):
            bottom = section * section_height
            ax.bar(x, section_height, bottom=bottom, width=0.76,
                   color=colors[outcome], edgecolor="white", linewidth=1.6, zorder=3)
            ax.text(x, bottom + section_height / 2, label,
                    ha="center", va="center", color="#243247", fontsize=9,
                    fontweight="medium")
        ax.annotate(f'{row["loss"]:.2f}', (x, row["loss"]),
                    xytext=(0, 8), textcoords="offset points", ha="center",
                    va="bottom", fontsize=12, fontweight="bold", color="#243247")

    ax.set_xticks(range(len(records)), [row["group"] for row in records])
    ax.set_ylabel("Avg training loss", labelpad=12)
    ax.set_ylim(0, max(row["loss"] for row in records) * 1.14)
    ax.set_xlim(-0.6, len(records) - 0.4)
    ax.set_axisbelow(True)
    ax.yaxis.grid(True, color="#E5E7EB", linewidth=0.8)
    ax.tick_params(axis="both", length=0, pad=9)
    for spine in ["top", "right", "left"]:
        ax.spines[spine].set_visible(False)
    ax.spines["bottom"].set_color("#CBD5E1")
    fig.suptitle("Training loss and hypothesis outcomes", x=0.085, y=0.96,
                 ha="left", fontsize=19, fontweight="bold", color="#182638")
    handles = [Patch(facecolor=color, label=status.capitalize())
               for status, color in colors.items()]
    fig.legend(handles=handles, loc="lower center", bbox_to_anchor=(0.5, 0.075),
               ncol=3, frameon=False)
    fig.subplots_adjust(left=0.085, right=0.98, top=0.82, bottom=0.20)
    plt.show()
```
